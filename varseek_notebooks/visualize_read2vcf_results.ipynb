````python
import pandas as pd
import matplotlib.pyplot as plt
from upsetplot import generate_counts, UpSet

def make_upset(data: dict[str, int], title=None, out_path=None, show=True) -> None:
    """
    data example:  {"both": 10, "Mutect2": 5, "varseek": 20}
    """
    if len(data) != 3:
        raise ValueError("data must have exactly three keys: 'both', 'Mutect2', and 'varseek'")
    if "both" not in data:
        raise ValueError("data must contain the key 'both'")
    
    keys = [k for k in data.keys() if k != "both"]

    # Define your overlap counts
    data_upset = [
        ((True, True), data["both"]),     # Both Mutect2 and Varseek
        ((True, False), data[keys[0]]),     # Mutect2 only
        ((False, True), data[keys[1]]),    # Varseek only
    ]

    # Create the Series with a MultiIndex and name
    example_counts2 = pd.Series(
        [count for _, count in data_upset],
        index=pd.MultiIndex.from_tuples([idx for idx, _ in data_upset], names=[keys[0], keys[1]]),
        name="value"
    )

    ax_dict = UpSet(example_counts2, show_counts=True).plot()
    if title:
        ax_dict["intersections"].set_title(title, fontsize=14, pad=15)
    if out_path:
        plt.savefig(out_path, bbox_inches="tight")
    if not show:
        plt.close()

def plot_sensitivity_and_ppv(data: dict[str, dict[str, float]], colors: dict[str, str] = None, length=6, out_path=None, show=True) -> None:
    """
    data example: {"A": {"sensitivity": 0.8, "ppv": 0.5}, "B": {"sensitivity": 0.7, "ppv": 0.6}, "C": {"sensitivity": 0.9, "ppv": 0.4}}
    """
    if colors is None:
        colors = {"sensitivity": "tab:blue", "ppv": "tab:orange"}

    for key, value in data.items():
        if not (isinstance(value, dict) and "sensitivity" in value and "ppv" in value):
            raise ValueError(f"Value for key '{key}' must be a dict with 'sensitivity' and 'ppv' keys")

    # Extract x (keys) and y values
    x = list(data.keys())
    y1 = [v["sensitivity"] for v in data.values()]
    y2 = [v["ppv"] for v in data.values()]

    # Plot
    plt.figure(figsize=(length,4))
    plt.plot(x, y1, label="Sensitivity", marker='o', color=colors["sensitivity"])
    plt.plot(x, y2, label="PPV", marker='o', color=colors["ppv"])

    # range go from 0-1, with grid lines every 0.1
    plt.ylim(0, 1)
    plt.yticks([i/10 for i in range(11)])
    plt.ylabel("Value")
    plt.legend(loc='upper left')
    plt.grid(True, linestyle="--", alpha=0.5)
    plt.tight_layout()
    if out_path:
        plt.savefig(out_path, bbox_inches="tight")
    if show:
        plt.show()
    plt.close()
````

````python
data = {
    "bcftools view": {"sensitivity": 0.7003, "ppv": 0.0184, "upset": {"both": 103319, "Mutect2": 44215, "varseek": 5508171}},
    "bcftools call, p=1e-3": {"sensitivity": 0.6860, "ppv": 0.2321, "upset": {"both": 101205, "Mutect2": 46329, "varseek": 334764}},
    "bcftools call, p=1e-2": {"sensitivity": 0.6889, "ppv": 0.2249, "upset": {"both": 101630, "Mutect2": 45904, "varseek": 350343}},
    "bcftools call, p=1e-1": {"sensitivity": 0.6908, "ppv": 0.2154, "upset": {"both": 101923, "Mutect2": 45611, "varseek": 371215}}
}
out_dir = "."

plot_sensitivity_and_ppv(data, length=6)
for key, value in data.items():
    make_upset(value["upset"], title=key, out_path=f"{out_dir}/{key}.pdf")
````
